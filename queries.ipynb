```python
from rdflib import Graph

# Ontolojiyi yükle
ontology_path = "/Users/gokcesoylu/ontology_engineering/ontology.owl"
g = Graph()
g.parse(ontology_path, format="xml")

# SPARQL sorgularını tanımla
queries = [
    # 1. Ontolojideki tüm sınıfları listele
    """
    SELECT DISTINCT ?class
    WHERE {
        ?class a owl:Class.
    }
    """,
    # 2. Ontolojideki bireylerin (individuals) listesini getir
    """
    SELECT DISTINCT ?individual
    WHERE {
        ?individual a ?type.
        FILTER(?type != owl:Class && ?type != owl:ObjectProperty).
    }
    """,
    # 3. Ontolojide tanımlı tüm object properties'i listele
    """
    SELECT DISTINCT ?property
    WHERE {
        ?property a owl:ObjectProperty.
    }
    """,
    # 4. Belirli bir sınıfın (örneğin "Person") bireylerini listele
    """
    SELECT DISTINCT ?individual
    WHERE {
        ?individual a <http://www.example.org#Person>.
    }
    """,
    # 5. İki birey arasındaki ilişkileri listele
    """
    SELECT DISTINCT ?subject ?predicate ?object
    WHERE {
        ?subject ?predicate ?object.
        FILTER(isIRI(?subject) && isIRI(?object)).
    }
    """
]

# Her bir sorguyu çalıştır ve sonuçları yazdır
for i, query in enumerate(queries, start=1):
    print(f"\nSorgu {i}:")
    results = g.query(query)
    for row in results:
        print(row)
```
