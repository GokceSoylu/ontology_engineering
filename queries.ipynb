```python
from rdflib import Graph, Namespace

# Ontoloji dosyasının yolu
ontology_path = "/Users/gokcesoylu/ontology_engineering/LastOntology.owl"

# RDFLib ile grafiği yükle
g = Graph()
g.parse(ontology_path, format="xml")

# Namespace tanımlama
ns = Namespace("http://example.org/ontology")  # Ontolojinizin doğru namespace'ini kullanın

# SPARQL sorgusu
query = """
SELECT DISTINCT ?eventType
WHERE {
  ?event a :Event .
  ?event :hasEventType ?eventType .
}
"""

# Sorguyu çalıştır ve sonuçları yazdır
for row in g.query(query, initNs={"": ns}):
    print(f"Event Type: {row.eventType}")
```

```python
from rdflib import Graph, Namespace
from rdflib.namespace import RDF, RDFS, OWL, XSD

# Load the ontology file
ontology_path = '/Users/gokcesoylu/ontology_engineering/LastOntology.owl'
g = Graph()
g.parse(ontology_path, format='xml')

# Define namespaces
ONTOLOGY = Namespace('http://example.org/ontology/')

# Query 1: Find all events and their locations
query_1 = """
SELECT ?event ?locationLabel
WHERE {
  ?event ontology:hasLocation ?location .
  ?location rdfs:label ?locationLabel .
}

"""

# Query 2: Find all participants of a specific event
query_2 = """
SELECT ?participant
WHERE {
    ?event rdf:type ontology:AcademicEvent .
    ?event ontology:hasParticipant ?participant .
}
"""

# Query 3: List all events organized by a specific organizer
query_3 = """
SELECT ?event ?label
WHERE {
  ?event ontology:hasOrganizer ontology:CodeConvene .
  ?event rdfs:label ?label .
}

"""

# Query 4: Find all events that occur in a specific city
query_4 = """
SELECT ?event ?label
WHERE {
  ?event ontology:hasLocation ontology:NewYork .
  ?event rdfs:label ?label .
}

"""

# Query 5: Infer participants who are also organizers
query_5 = """
SELECT ?participant ?organizer
WHERE {
  ?participant rdf:type ontology:PARTICIPANT .
  ?organizer rdf:type ontology:ORGANIZER .
  FILTER (?participant = ?organizer)
}

"""

# Execute and display the results for each query
queries = [query_1, query_2, query_3, query_4, query_5]

for i, query in enumerate(queries, 1):
    print(f"Results for Query {i}:")
    results = g.query(query)
    for row in results:
        print(row)
    print("\n")
```
